# Bilan de masse cumulé

Valeurs illustratives du Glacier-Exemple. Exécute les cellules une à une (Maj+Entrée).

```python
import csv

annees, bilans = [], []
with open("donnees/bilan_masse_annuel.csv", newline="") as f:
    for ligne in csv.DictReader(f):
        annees.append(int(ligne["annee"]))
        bilans.append(float(ligne["bilan_m_eq_eau"]))

cumule = []
total = 0.0
for b in bilans:
    total += b
    cumule.append(round(total, 2))
print(f"Perte cumulée 2000-2024 : {cumule[-1]} m éq. eau")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(annees, cumule, color="#3b6e99")
ax.set_xlabel("Année")
ax.set_ylabel("Bilan cumulé (m éq. eau)")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
